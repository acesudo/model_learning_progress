## now we make our own class GDregressor 

```python
from sklearn.datasets import make_regression
```

```python
X,y=make_regression(n_samples=100,n_features=1,n_targets=1,noise=20)
```

```python
import matplotlib.pyplot as plt 
```

```python
plt.scatter(X,y)
plt.show()
```

```python
## just for reference we will first see the Linear regression 

from sklearn.linear_model import LinearRegression

lr = LinearRegression()
lr.fit(X,y)
```

```python
y_predict = lr.predict(X)

print("the slope by ols method :",lr.coef_)
print("the intercept by ols:",lr.intercept_)

plt.scatter(X,y)
plt.plot(X,y_predict ,color="red")
```

## lets make GDregressor class assuming m is 68.52 means constant

```python
import numpy as np 
```

```python
class GDregressor:
    def __init__(self,learning_rate,epoches):
        self.m =68.52
        self.lr=learning_rate
        self.epoches=epoches
        self.b=0
    def fit(self,X,y):
        for i in range(self.epoches):
            loss_slope= -2*np.sum(y-self.m*X.ravel()-self.b)
            self.b=self.b-(self.lr*loss_slope)
        print(self.b)    
        
```

```python
reg =GDregressor(0.001,10)
```

```python
reg.fit(X,y)
```
