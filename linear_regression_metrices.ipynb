## linear regression metrices for model 

```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\hp\Music\ml_classproject\practise_folder\cgpa_package_realistic_150.csv")

df.head(10)
```

```python
plt.scatter(df["CGPA"] ,df["Package_LPA"])
plt.xlabel("cgpa")
plt.ylabel("package")
plt.show()
```

```python
X_train_value = df[["CGPA"]]
Y_train_value = df[["Package_LPA"]]
```

```python
from sklearn.model_selection import train_test_split

X_train,x_test,Y_train,y_test = train_test_split(X_train_value,Y_train_value,test_size=0.2,random_state=2)
```

```python
from sklearn.linear_model import LinearRegression

lr = LinearRegression()
lr.fit(X_train,Y_train)
```

```python
y_predict=lr.predict(x_test)
```

```python
plt.scatter(df["CGPA"],df["Package_LPA"])
plt.plot(x_test,y_predict ,color="red")
```

## linear regression metrices


```python
## mean absolute error 

from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score

print("mean absolute error of model prediction in lpa  ", mean_absolute_error(y_test,y_predict))
```

```python
print("mean squared error in lpa^2 :" , mean_squared_error(y_test,y_predict))
```

```python
print("r2 score :", r2_score(y_test,y_predict))
r2=r2_score(y_test,y_predict)
```
