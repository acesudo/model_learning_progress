```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np 
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
```

```python
data = pd.read_csv('polynomial_regression_dataset.csv')

data.head(10)
```

```python
plt.plot(data["Hours_Studied"],data["Exam_Score"], 'b.')
plt.xlabel("hours_studied")
plt.ylabel("Exam_scored")
plt.show()
```

## in this polynomial dataset we fir st try to fit simple linear regression and see whaat happened


```python
X_train_value =data[["Hours_Studied"]]
y_train_value =data[["Exam_Score"]]
```

```python
X_train,x_test,Y_train,y_test = train_test_split(X_train_value,y_train_value ,test_size=0.2,random_state=2)
```

```python
lr=LinearRegression()
lr.fit(X_train,Y_train)
```

```python
y_predicted_value= lr.predict(x_test)
```

## now we train our model lets see its graphical represenation 


```python
plt.scatter(data["Hours_Studied"],data["Exam_Score"])
plt.plot(x_test,y_predicted_value,color="red")
```

```python
# we will justified it by using r2 score

r2= r2_score(y_test,y_predicted_value)
r2
```
