```python
# donut chart is basically pie chart witha huge hole in the center . matplotlib and seaborn do not have a direct function for it . so we need to create one 
x=[20,30,10,70,60]
y=["A","B","C","D","E"]

# create a pie chart with a hole in it by wedgeprops to make a hole
plt.pie(x,labels=y,autopct="%1.2f%%",wedgeprops={"width":0.4},explode=[0.3,0.2,0,0,0])
plt.show()
```

**Problem Statement**
A retail company wants to analyze the quarterly sales of three products (Laptop, Mobile, Tablet) over 4 quarters of the year 2023.
The goal is to compare product sales within each quarter and also see how sales change across quarters.

```python
lp_sale=[233,343,300,890]
mp_sale=[230,311,201,680]
tb_sale=[345,456,564,600]

X=["Q1","Q2","Q3","Q4"]

xx=np.arange(len(X))
width=0.75
plt.bar(xx-width,lp_sale)
plt.bar(xx,mp_sale)
plt.bar(xx+width,tb_sale)
```

```python
# stacked bar chart  astacked bar chart is a type of bar chart where categories where stacked on top of each other in a single bar , insead of placing them side by side 
# like in rouping or clustering bar chart 
# the length of the full bar represent the main categories 
# other than main section , each section represent the subcategories

categories = ['Q1', 'Q2', 'Q3', 'Q4']
laptop_sales = [200, 250, 300, 280]
mobile_sales = [220, 270, 320, 310]
tablet_sales = [150, 180, 200, 190]

x = np.arange(len(categories))

plt.bar(x,laptop_sales)
plt.bar(x,mobile_sales,bottom=laptop_sales)
plt.bar(x,tablet_sales,bottom=(np.array(laptop_sales)+np.array(mobile_sales)))

plt.xticks(x,categories)
plt.legend(["laptop","mobile","tablet"])
```

```python
categories = ['Q1', 'Q2', 'Q3', 'Q4']
laptop_sales = [200, 250, 300, 280]
mobile_sales = [220, 270, 320, 310]
tablet_sales = [150, 180, 200, 190]

x = np.arange(len(categories))  # [0, 1, 2, 3]

# Plot stacked bars
plt.bar(x, laptop_sales, label="Laptop", color="skyblue")
plt.bar(x, mobile_sales, bottom=laptop_sales, label="Mobile", color="lightgreen")
plt.bar(x, tablet_sales, bottom=np.array(laptop_sales) + np.array(mobile_sales),
        label="Tablet", color="salmon")

# Add data labels
for i in range(len(x)):
    # Laptop labels
    plt.text(x[i], laptop_sales[i] / 2, laptop_sales[i], ha="center", va="center", color="black")
    
  # plt .text(x_positiin,y_position,text)

    # Mobile labels
    plt.text(x[i], laptop_sales[i] + mobile_sales[i] / 2, mobile_sales[i], ha="center", va="center", color="black")

    # Tablet labels
    plt.text(x[i], laptop_sales[i] + mobile_sales[i] + tablet_sales[i] / 2,
             tablet_sales[i], ha="center", va="center", color="black")

# Labels
plt.xlabel("Quarter")
plt.ylabel("Sales")
plt.title("Stacked Bar Chart: Product Sales by Quarter (2023)")
plt.xticks(x, categories)
plt.legend()
plt.grid(axis="y", linestyle="--", alpha=0.7) # horizontal grid

plt.show()
```

```python
# second method is fig , axes in subplots

# Data
months = [1, 2, 3, 4, 5]
product_X = [12, 18, 22, 15, 30]
product_Y = [8, 14, 19, 25, 28]
product_Z = [20, 25, 27, 22, 26]
product_W = [5, 10, 8, 15, 18]

# method second
fig,axis=plt.subplots(2,2)

axis[0,0].plot(months,product_X,label="X")
axis[0,0].set_title("product_X")
axis[0,0].grid()

axis[0,1].plot(months,product_Y,label="Y")
axis[0,1].set_title("product_y")
axis[0,1].grid()

axis[1,0].plot(months,product_Z,label="Z")
axis[1,0].set_title("product_Z")
axis[1,0].grid()

axis[1,1].plot(months,product_W,label="W")
axis[1,1].set_title("product_W")

axis[1,1].grid()

plt.tight_layout()
plt.legend()
```

```python
# Arear chart is a line chart but the space between the line is filled with color, its shows a how a value changed over time or categories 
# use ful for visualizing trends and magnitude

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [100, 150, 200, 180, 220, 300]

plt.plot(months,sales,color="blue")
plt.fill_between(months,sales,color="skyblue",alpha=0.5)
```

```python
# area chart for two product comparision 
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
product_A = [100, 150, 200, 180, 220, 300]
product_B = [80, 120, 160, 140, 180, 240]

plt.plot(months,product_A,color="blue")
plt.fill_between(months,product_A,color="skyblue",alpha=0.5)

plt.plot(months,product_B,color="orange")
plt.fill_between(months,product_B,color="pink",alpha=0.5)

plt.title("productA vs product B area chart ",fontsize=12,fontweight="bold")
plt.xlabel("months")
plt.ylabel("prduct_sales")
plt.grid(linestyle="--",alpha=0.7)
plt.legend(["product A","product B"])
```
