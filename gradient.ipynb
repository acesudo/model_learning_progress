## gradient desecnt in machine learning optimisation technique

```python
from sklearn.datasets import make_regression
import numpy as np
```

```python
X,y = make_regression(n_samples=4,n_features=1,n_informative=1,n_targets=1,noise=80,random_state=13)
```

```python
import matplotlib.pyplot as plt 
```

```python
plt.scatter(X,y)
plt.show()
```

```python
# linear regression method which use ols in background 

from sklearn.linear_model import LinearRegression
lr1= LinearRegression()
lr1.fit(X,y)
```

```python
lr1.coef_
```

```python
lr1.intercept_
```

```python
plt.scatter(X,y)
plt.plot(X,lr1.predict(X),color="red")
plt.show()
```

## now we use gradient descent assuming m= 78.35 and b=0

```python
m=78.35
b=0

loss_slope = -2 * np.sum(y-m*X.ravel()-b)
loss_slope
```

```python
# initial value of y predicted value when b =0

y_predict = ((78.35*X)+0).reshape(4)
```

```python
# now learning rate 
lr=0.1
step_size=loss_slope*lr
step_size
```

```python
b=b-step_size
b
```

```python
# y predicted for new value of b 
y1_predict =((78.35*X)+b).reshape(4)
```

```python
plt.scatter(X,y)
plt.plot(X,lr1.predict(X) ,color="black")
plt.plot(X,y_predict,color="pink")
plt.plot(X,y1_predict,color="red")
```

## hence wtih every iteration we are getting close and close to the actaul line predicted by ols method
