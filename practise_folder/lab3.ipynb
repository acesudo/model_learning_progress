```python
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
import scipy.stats as stats
```

```python
df =pd.read_csv(r"C:\Users\hp\Music\ml_classproject\all_csv_folder\student_placement_dataset.csv")
df
```

```python
df.describe()
```

```python
# remove duplicated 
df=df.drop_duplicates()

df.info()
```

```python
# fill the missing values
df.isnull().sum()
```

```python
# for cgpa we first find out he which we use 
sns.distplot(df["CGPA"])
# this column does not contain any swekness soo we can use mean for filling the null values
```

```python
stats.probplot(df["CGPA"] , dist="norm" ,plot=plt)
```

```python
df["CGPA"]=df["CGPA"].fillna(df["CGPA"].mean())

df["CGPA"].isnull().sum()
```

```python
# we will check the aptitude score 
df["AptitudeScore"].hist(bins=30 , edgecolor="black")
```

```python
sns.distplot(df["AptitudeScore"])
```

```python
df["AptitudeScore"]=df["AptitudeScore"].fillna(df["AptitudeScore"].mean())

df["AptitudeScore"].isnull().sum()
```

```python
# technical and communication score
sns.distplot(df["TechnicalScore"])
sns.distplot(df["CommunicationScore"])
```

```python
df["TechnicalScore"]=df["TechnicalScore"].fillna(df["TechnicalScore"].mean())
df["CommunicationScore"]=df["CommunicationScore"].fillna(df["CommunicationScore"].mean())

print(df["CommunicationScore"].isnull().sum())
print(df["TechnicalScore"].isnull().sum())
```

```python
# for intership cloumn
# since data is categorica; we use the mode operations
df["Internship"]=df["Internship"].fillna(df["Internship"].mode()[0])

df["Internship"].isnull().sum()
```
