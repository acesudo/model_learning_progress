Handling missing values

```python
import numpy as np
import pandas as pd
```

```python
d={
    "name":["kunal","aisha","midoriya","rahul","ronakita"],
    "age":[12,np.nan,np.nan,13,14],
    "subject":["maths","english","maths","english",np.nan]
}
d1=pd.DataFrame(d)
d1
```

```python
d1.isna()  # it give true if any cell contain null values and gives not if it not contain null value
```

```python
d1.isna().sum()   # it count null value in each column
```

```python
d1.dropna(thresh=3)  # it means atleast three non null value must be present in a row
                      # here thresh mean threshold
```

```python
d1.dropna(axis=1)  # dropna mean drop all the rows or column containing null values
d1.dropna(axis=0)  # here in dropna axis --> 0 means drop row-wise and axis--->1 mean drop column-wise
```

```python
d1.fillna(0)  # it will fill all the null value with given bracket value  and it select column wise
```

```python
d1.fillna(1,axis=1)  # we can use axis for filling rows or column wise
```

```python
d1
```

```python
values={"age":12,"subject":"painting"}  # we can fill anything column wise
d1.fillna(value=values)
```

merging and concatination in dataframe

```python
# concatination means copy and pasting two dataframe with each other
data_2={
    "a":[1,2,3,4],
    "b":[5,6,7,8],
    "c":[9,10,11,12]
}
d2=pd.DataFrame(data_2)
data_3={
    "a":[13,14,15,16],
    "b":[17,18,19,20],
    "c":[21,22,23,24]
}
d3=pd.DataFrame(data_3)
```

```python
d2
```

```python
con_1=pd.concat([d2,d3],axis=1)   # concatination done by rows -wise
con_1.columns=["a1","b1","c1","a2","b2","c2"]
con_1
```

```python
con_1=pd.concat([d2,d3],axis=0)
con_1.index=[0,1,2,3,4,5,6,7]     # we concat two dataframe with index - wise
```

```python
# joining   series to form  dataframe means dealing dataframe based on their index axis matlab konsi dono me hai or konsi nhi
```

```python
d4=pd.DataFrame({"a":[1,2,3,4],"b":[11,22,33,44],"c":[111,222,333,444]},index=[1,2,3,4])
d5=pd.DataFrame({"a1":[5,6,7,8],"b1":[55,66,77,88],"c1":[555,666,777,888]},index=[4,5,6,7])

d4.join(d5,how="outer")
```

groupby and aggeragation


```python
d6=pd.DataFrame({"category":["s1","s2","s1","s3","s1","s3","s1","s2","s1","s1","s3","s1"],
                 "store":["A","C","A","B","C","B","D","B","A","B","D","B"],
                 "rate":[100,200,100,300,200,400,500,100,600,900,8,9]})
d6
```

```python
# groupby method return objcet of same group which we can see by using for loop
d6.groupby(['category','store']).sum()
```

```python
# less begin with aggregation function soo less go
```

```python
d6
```

```python
print(d6['rate'].sum())  # these are some aggeregation function
print(d6['rate'].mean())
print(d6['rate'].min())
```

```python
# aggeragation mean operations perform krna kisisbhi column or rows pr

d6['rate'].agg(['min','mean','sum'])  # yup we can provide all operation in alist to do it effectively and at a same time
```
