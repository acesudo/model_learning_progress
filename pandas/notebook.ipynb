DATA FILTERINGPROCESS 

```python
data=pd.DataFrame({
    "Name":["ha","pi","jin","pin"],
    "age":[12,13,14,15],
    "gender":["male","female","male","female"],
    "quali.":["12th","11th","10th","9th"],
    })
df=pd.DataFrame(data)
df
```

```python
# renaming the column 
df.rename(columns={"quali.":"Qualification"},inplace=True)
print(df.columns)
```

```python
# selecting of particular data
df
```

```python
# using operator 
df[df["Name"]=="ha"]
```

```python
df
```

```python
# for multiple condition we have to use the () brackets 
df[(df["Name"]=="ha") & (df["age"] >= 10)]
```

```python
# we can also find speceifc rows and column by string method
df[df["Name"].str.startswith("p")][["gender","Qualification"]]
```

```python
# we can also filter the data by using isin method
df
```

```python
selected_gender=["female"]
df[df["gender"].isin(selected_gender)]
```

```python
# we cna also use iloc for filtering using index and loc for filtering using labels
df
```

```python
df.iloc[0:2] 
```

```python
df.loc["0":"3"]
```

IDENTIFYING AND HANDLING MISSING VALUES 

```python
import numpy as np
data=pd.DataFrame({
    "name":["kunal","rangi","pesi","suii","messi","rolando","msd"],
    "age":[17,14,19,41,39,41,45],
    "job":[np.nan,np.nan,"youtuber","footbaleer","footballer","footballer","cricketor"],
    "class":[np.nan,np.nan,np.nan,np.nan,np.nan,np.nan,np.nan]
})
data
```

```python
# to find null value in a dataframe    
data.isnull()    # it give boolen value column wise true = null value and false= nt null value
```

```python
# we can count total no .of null value in every clomun
data.isnull().sum()    
```

```python
# if one column contain  null value we can drop it 
data.dropna()
```

```python
data
```

```python
data.dropna(axis=1)  # we can drop values both column and rows wise by using 'axis'
```

```python
data.dropna(thresh=3, axis=1)  # it mean if minimum 3 nt null value are present it will not drop (default it work row-wise)
```

```python
# we can fill the null value 
data
```

```python
data.fillna(0)  # fill nan vlaue with 0
```

```python
data["class"]=data["class"].fillna(0)
data
```

```python
# we can select multiple column and give each of them a seperate value
data
```

```python
missing_values={"class":10,"job":"unemployed"}
data.fillna(missing_values)
```
